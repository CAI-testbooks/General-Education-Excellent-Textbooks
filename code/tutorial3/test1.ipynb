```python
import os
import requests
from bs4 import BeautifulSoup
```

```python
import urllib.parse
```

```python
def download_snow_images(num_images=50, save_dir='snow_images'):
    # 创建保存图片的目录
    if not os.path.exists(save_dir): os.makedirs(save_dir)
    # Bing图片搜索的URL
    search_term = 'snow street real-world person auto-driven'
    url = f'https://www.bing.com/images/search?q={urllib.parse.quote(search_term)}&first=1&tsc=ImageBasicHover'

    # 发送GET请求
    response = requests.get(url)
    soup = BeautifulSoup(response.text, 'html.parser')

    # 查找所有图片链接
    img_tags = soup.find_all('img', class_='mimg')

    # 下载图片
    count = 0
    for img in img_tags:
        if count >= num_images:
            break

        img_url = img.get('src')
        if img_url and img_url.startswith('http'):
            try:
                img_data = requests.get(img_url).content
                with open(os.path.join(save_dir, f'snow_image_{count}.jpg'), 'wb') as handler:
                    handler.write(img_data)
                print(f'Downloaded image {count + 1}')
                count += 1
            except Exception as e:
                print(f'Error downloading image: {e}')
    print(f'Downloaded {count} images to {save_dir}')
```

```python
# 调用函数下载50张雪天图片
download_snow_images(10)
```
